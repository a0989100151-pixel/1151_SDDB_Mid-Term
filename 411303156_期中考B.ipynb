## 3. 資料表（2 張）
| 表名 | 欄位 | 型態 | 說明 |
|---|---|---|---|
| menu（菜單） | item_id | INTEGER（主鍵、自動編號） | 品項編號 |

初始資料：menu 三筆——美式咖啡 單價 50 庫存 5；拿鐵 單價 65 庫存 4；抹茶歐蕾 單價 70 庫存 3。orders 一開始是空的。每次執行都先 DROP TABLE IF EXISTS 再重建。

剩餘杯數的算法：剩餘 ＝ 今日庫存 − 該品項所有「已成立」訂單的杯數加總。取消訂單只要把狀態改成「已取消」，庫存就自動釋出。

## 4. AI 的工作
- 輸入：使用者的一句話
- 輸出格式：`動作,品項,數量`
- 動作只能是：訂購、取消、查詢、統計（各品項銷售狀況）
- 品項只能是：美式咖啡、拿鐵、抹茶歐蕾
- 數量的規則：數量是杯數，只輸出阿拉伯數字（「兩杯」＝2、「十杯」＝10）；沒說就是 1；AI 照實輸出，不自己限制範圍，數量是否在 1～3 之間由程式檢查
- 無法判斷時輸出：`未知,無,1`（品項不存在、沒說品項、或同時提到多個品項時，品項輸出「無」）；系統收到後會回覆目前有哪些品項
- AI 不可以做的事：不可以自己執行訂購或取消、不可以回答問題或輸出說明文字（只能輸出一行）、不可以編造不存在的品項、不可以判斷庫存夠不夠或數量合不合法。程式也不直接相信 AI 的回覆，只從回覆中找出允許清單裡的值：找不到動作就當「未知」、找不到品項就當「無」、找不到數字就當 1

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# 這一格做三件事：
# 1. 設定資料庫檔名與固定的常數（動作清單、品項清單、數量範圍、每人上限）
# 2. 定義 init_db()：刪掉舊表 → 建立「菜單」「訂單」兩張新表 → 放入初始資料
# 3. 執行 init_db()，確保每次「全部執行」都是乾淨的狀態
#
# 設計重點：menu.stock 存的是「今日庫存（初始杯數）」，不會被修改；
# 剩餘杯數 = 今日庫存 − 所有「已成立」訂單的杯數加總。
# 這樣取消訂單只需要把狀態改成「已取消」，庫存就自動釋出，不容易算錯。

DB_PATH = "cafe.db"

ACTIONS = ["訂購", "取消", "查詢", "統計"]        # AI 只能輸出這幾種動作（無法判斷時為「未知」）
ITEMS = ["美式咖啡", "拿鐵", "抹茶歐蕾"]            # AI 只能輸出這幾個品項（找不到時為「無」）
MIN_QTY, MAX_QTY = 1, 3                            # 單次訂購的數量範圍
MAX_PER_STUDENT = 3                                # 每位學生今天最多訂幾杯（已取消的不計）

INITIAL_MENU = [
    ("美式咖啡", 50, 5),
    ("拿鐵", 65, 4),
    ("抹茶歐蕾", 70, 3),
]


def get_conn():
    """建立資料庫連線（每次用完要記得關閉）"""
    return sqlite3.connect(DB_PATH)


def init_db():
    """重建兩張資料表並放入初始資料"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        # 先刪「訂單」再刪「菜單」，因為訂單會參考菜單
        cur.execute("DROP TABLE IF EXISTS orders")
        cur.execute("DROP TABLE IF EXISTS menu")

        cur.execute("""
            CREATE TABLE menu (
                item_id INTEGER PRIMARY KEY AUTOINCREMENT,
                name    TEXT NOT NULL UNIQUE,
                price   INTEGER NOT NULL,
                stock   INTEGER NOT NULL
            )
        """)
        cur.execute("""
            CREATE TABLE orders (
                order_id   INTEGER PRIMARY KEY AUTOINCREMENT,
                student    TEXT NOT NULL,
                item_id    INTEGER NOT NULL,
                quantity   INTEGER NOT NULL,
                status     TEXT NOT NULL DEFAULT '已成立',
                created_at TEXT NOT NULL DEFAULT CURRENT_TIMESTAMP,
                FOREIGN KEY (item_id) REFERENCES menu(item_id)
            )
        """)

        cur.executemany(
            "INSERT INTO menu (name, price, stock) VALUES (?, ?, ?)",
            INITIAL_MENU,
        )
        conn.commit()
    finally:
        conn.close()


init_db()
print("✅ 資料庫已重建，初始資料已放入")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# show_tables() 會把兩張資料表的內容用表格顯示出來，方便檢查資料庫有沒有改對。

def show_tables():
    """顯示 menu 與 orders 兩張表的全部內容"""
    conn = get_conn()
    try:
        menu_df = pd.read_sql_query("SELECT * FROM menu", conn)
        orders_df = pd.read_sql_query("SELECT * FROM orders", conn)
    finally:
        conn.close()

    print("🍵 menu（菜單；stock＝今日庫存）")
    display(menu_df)
    print("🧾 orders（訂單）")
    if orders_df.empty:
        print("（目前沒有任何訂單）")
    else:
        display(orders_df)


show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# 流程：使用者一句話 → 呼叫 Groq AI → 拿到「動作,品項,數量」→
# 程式自己檢查 AI 的回覆（只接受允許清單裡的值）→ 回傳整理好的結果。
# 重點：AI 只負責「讀懂句子」，數量合不合法由程式判斷，所以 AI 要照實輸出數字。

SYSTEM_PROMPT = """你是「校園咖啡廳預約系統」的文字解讀器，只負責把使用者的一句話轉成一行固定格式。
輸出格式（只能輸出這一行，不要加任何說明、標題、標點、空白行或引號）：
動作,品項,數量

規則：
1. 動作只能是：訂購、取消、查詢、統計、未知。
   - 訂購：想買、想訂、想點飲料。 取消：想退掉、不要了。 查詢：問某品項還剩幾杯。
   - 統計：想看「所有品項」的銷售狀況、賣了多少、營收（只輸出一行，不要逐項列出）。
   - 無法判斷或與咖啡廳無關時輸出：未知
2. 品項只能是：美式咖啡、拿鐵、抹茶歐蕾。使用者說的品項不在這三個之中、沒有說品項、或同時提到多個品項時，品項輸出：無
   統計時品項輸出：無
3. 數量是杯數，只輸出阿拉伯數字。「兩杯」＝2、「三杯」＝3、「十杯」＝10。沒有提到數量就輸出 1。
   請照使用者說的數字照實輸出，不要自己限制範圍（即使是 0 或 10 也照實輸出），數量合不合理由程式判斷。
4. 你不可以執行訂購或取消，不可以回答問題，不可以編造不存在的品項。

範例：
我要訂一杯美式咖啡 → 訂購,美式咖啡,1
來兩杯拿鐵 → 訂購,拿鐵,2
給我十杯抹茶歐蕾 → 訂購,抹茶歐蕾,10
幫我取消拿鐵 → 取消,拿鐵,1
抹茶歐蕾還剩幾杯 → 查詢,抹茶歐蕾,1
今天各品項賣得怎樣 → 統計,無,1
今天營收多少 → 統計,無,1
我要訂一杯珍珠奶茶 → 訂購,無,1
今天天氣怎麼樣 → 未知,無,1
"""

AI_CACHE = {}   # 同一句話只問一次 AI，成功的結果暫存在這裡（節省額度）


def call_ai(text):
    """呼叫 Groq，回傳 AI 的原始文字；失敗時回傳空字串。
    遇到 429（用量超過）等 30 秒再試，最多 3 次。"""
    for attempt in range(3):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                max_completion_tokens=500,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": text},
                ],
            )
            time.sleep(2)   # 每次呼叫後休息 2 秒，避免太快用完額度
            return (resp.choices[0].message.content or "").strip()
        except Exception as e:
            time.sleep(2)
            if "429" in str(e) or getattr(e, "status_code", None) == 429:
                print(f"⏳ 超過使用量，等待 30 秒後重試（第 {attempt + 1}/3 次）")
                time.sleep(30)
            else:
                print("⚠️ 呼叫 AI 失敗：", type(e).__name__)
                return ""
    return ""


def parse_ai_reply(raw):
    """不直接相信 AI：自己從回覆中找出允許清單裡的值。找不到就用預設值。"""
    # AI 偶爾會多回幾行（標題、說明），只看「第一行含有合法動作」的那一行
    lines = [ln for ln in raw.splitlines() if ln.strip()]
    line = next((ln for ln in lines if any(a in ln for a in ACTIONS)),
                lines[0] if lines else "")

    # 動作：找「最早出現」的允許動作，找不到＝未知
    action, best = "未知", None
    for a in ACTIONS:
        pos = line.find(a)
        if pos != -1 and (best is None or pos < best):
            action, best = a, pos

    # 品項：同樣找最早出現的允許品項，找不到＝無
    item, best = "無", None
    for it in ITEMS:
        pos = line.find(it)
        if pos != -1 and (best is None or pos < best):
            item, best = it, pos

    # 數量：優先看第 3 個欄位，用正規表示式找數字；找不到就當作 1
    parts = re.split(r"[,，]", line)
    source = parts[2] if len(parts) >= 3 else line
    m = re.search(r"\d+", source)
    qty = int(m.group()) if m else 1
    return action, item, qty


def interpret(text):
    """AI 解讀主函式：回傳 (動作, 品項, 數量, 原始回覆)"""
    key = text.strip()
    if key not in AI_CACHE:
        raw = call_ai(key)
        if not raw:                       # 呼叫失敗時不暫存，下次還可以重問
            return "未知", "無", 1, raw
        AI_CACHE[key] = raw
    raw = AI_CACHE[key]
    action, item, qty = parse_ai_reply(raw)
    return action, item, qty, raw


print("✅ AI 解讀函式已定義")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# 格式：(使用者說的話, 預期動作, 預期品項, 預期數量)，"*" 表示不在意
# 動作、品項、數量三個都會比對（有指定時）
AI_TESTS = [
    ("我要訂一杯美式咖啡", "訂購", "美式咖啡", 1),  # 1 正常
    ("幫我取消拿鐵", "取消", "拿鐵", "*"),  # 2 正常
    ("抹茶歐蕾還剩幾杯？", "查詢", "抹茶歐蕾", "*"),  # 3 正常
    ("今天各品項賣得怎樣", "統計", "*", "*"),  # 4 正常（統計）
    ("我要兩杯拿鐵", "訂購", "拿鐵", 2),  # 5 正常（數量）
    ("來三杯抹茶歐蕾", "訂購", "抹茶歐蕾", 3),  # 6 邊界（數量上限）
    ("給我十杯美式咖啡", "訂購", "美式咖啡", 10),  # 7 邊界（超過上限，由程式拒絕）
    ("我要訂一杯珍珠奶茶", "訂購", "無", "*"),  # 8 錯誤（品項不存在）
    ("今天天氣怎麼樣", "未知", "無", "*"),  # 9 錯誤（無關的話）
    ("我想訂飲料", "訂購", "無", "*"),  # 10 邊界（沒說品項）
]


def run_ai_tests():
    passed = 0
    for i, (text, exp_action, exp_item, exp_qty) in enumerate(AI_TESTS, start=1):
        action, item, qty, raw = interpret(text)
        ok = (
            action == exp_action
            and (exp_item == "*" or item == exp_item)
            and (exp_qty == "*" or qty == exp_qty)
        )
        passed += ok
        print(f"{i:>2}. {'✅' if ok else '❌'} 「{text}」")
        print(f"    AI 原始回覆：{raw!r}")
        print(f"    系統解讀：動作={action}、品項={item}、數量={qty}"
              f"（預期：{exp_action}、{exp_item}、{exp_qty}）")
    print(f"\n🎯 AI 解讀測試：通過 {passed} / {len(AI_TESTS)} 句")


run_ai_tests()
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# 這一格定義系統的所有功能：訂購、取消、查詢、統計，以及總入口 handle_request()。
# 剩餘杯數 = 今日庫存 − 「已成立」訂單的杯數加總（已取消的不算）。
# 只有「訂購、取消」需要姓名；「查詢、統計」不需要。

def _find_item(cur, name):
    """用品項名稱找品項，回傳 (item_id, 單價, 今日庫存)，找不到回傳 None"""
    cur.execute("SELECT item_id, price, stock FROM menu WHERE name = ?", (name,))
    return cur.fetchone()


def _sold(cur, item_id):
    """某品項目前已售出的杯數（只算「已成立」）"""
    cur.execute(
        "SELECT COALESCE(SUM(quantity), 0) FROM orders WHERE item_id = ? AND status = '已成立'",
        (item_id,),
    )
    row = cur.fetchone()
    return row[0] if row is not None else 0


def menu_text():
    """列出目前有哪些品項（AI 看不懂時用）"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("SELECT name, price FROM menu ORDER BY item_id")
        rows = cur.fetchall()
    finally:
        conn.close()
    return "\n".join(f"・{name}（{price} 元）" for name, price in rows)


def order_drink(student, item, qty):
    """訂購：檢查數量範圍 → 每人上限 → 庫存 → 寫入訂單，成功時回覆金額"""
    if qty < MIN_QTY or qty > MAX_QTY:
        return f"❌ 訂購失敗：單次數量需介於 {MIN_QTY}～{MAX_QTY} 杯，您輸入的是 {qty} 杯。"

    conn = get_conn()
    try:
        cur = conn.cursor()
        found = _find_item(cur, item)
        if found is None:
            return "❌ 查無此品項。"
        item_id, price, stock = found

        # 每位學生今天最多 3 杯（所有品項加總，已取消的不計）
        cur.execute(
            "SELECT COALESCE(SUM(quantity), 0) FROM orders WHERE student = ? AND status = '已成立'",
            (student,),
        )
        row = cur.fetchone()
        already = row[0] if row is not None else 0
        if already + qty > MAX_PER_STUDENT:
            return (f"❌ 訂購失敗：每人今天最多 {MAX_PER_STUDENT} 杯，"
                    f"{student} 已訂 {already} 杯，再訂 {qty} 杯會超過每人上限。")

        # 庫存是否足夠（用杯數計算）
        remaining = stock - _sold(cur, item_id)
        if qty > remaining:
            return f"❌ 訂購失敗：「{item}」庫存不足，只剩 {remaining} 杯，無法訂 {qty} 杯。"

        cur.execute(
            "INSERT INTO orders (student, item_id, quantity, status) VALUES (?, ?, ?, '已成立')",
            (student, item_id, qty),
        )
        conn.commit()
        return (f"✅ 預約成功！{student} 訂購 {qty} 杯「{item}」，"
                f"金額 {price * qty} 元，「{item}」剩餘 {remaining - qty} 杯。")
    finally:
        conn.close()


def cancel_order(student, item):
    """取消：把該學生該品項「最新一筆」已成立的訂單改成「已取消」，庫存自動釋出"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        found = _find_item(cur, item)
        if found is None:
            return "❌ 查無此品項。"
        item_id, price, stock = found

        cur.execute(
            "SELECT order_id, quantity FROM orders "
            "WHERE student = ? AND item_id = ? AND status = '已成立' "
            "ORDER BY order_id DESC LIMIT 1",
            (student, item_id),
        )
        row = cur.fetchone()
        if row is None:
            return f"⚠️ 查無訂單：{student} 沒有「{item}」的有效訂單，無法取消。"
        order_id, quantity = row

        cur.execute("UPDATE orders SET status = '已取消' WHERE order_id = ?", (order_id,))
        conn.commit()
        remaining = stock - _sold(cur, item_id)
        return f"✅ 取消成功！已釋出 {quantity} 杯，「{item}」剩餘 {remaining} 杯。"
    finally:
        conn.close()


def query_stock(item):
    """查詢：回覆某品項的剩餘杯數（不需要姓名）"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        found = _find_item(cur, item)
        if found is None:
            return "❌ 查無此品項。"
        item_id, price, stock = found
        remaining = stock - _sold(cur, item_id)
        return f"🔎「{item}」（{price} 元）今日庫存 {stock} 杯，剩餘 {remaining} 杯。"
    finally:
        conn.close()


def get_sales():
    """統計：JOIN + GROUP BY 列出每個品項的單價、庫存、已售杯數、剩餘杯數、營收。
    用 LEFT JOIN，而且「已成立」條件放在 ON 裡，這樣沒有人訂的品項也會列出來。"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("""
            SELECT m.name, m.price, m.stock, COALESCE(SUM(o.quantity), 0) AS sold
            FROM menu AS m
            LEFT JOIN orders AS o
                   ON o.item_id = m.item_id AND o.status = '已成立'
            GROUP BY m.item_id, m.name, m.price, m.stock
            ORDER BY m.item_id
        """)
        rows = cur.fetchall()
    finally:
        conn.close()

    lines = ["📊 今日銷售統計"]
    total = 0
    for name, price, stock, sold in rows:
        revenue = price * sold
        total += revenue
        lines.append(f"・{name}｜單價 {price}｜庫存 {stock}｜已售 {sold}｜剩餘 {stock - sold}｜營收 {revenue} 元")
    lines.append(f"合計營收：{total} 元")
    return "\n".join(lines)


def unknown_reply():
    return "🤔 我看不懂您的需求，或是沒有這個品項。目前有以下品項：\n" + menu_text()


def handle_request(text, student):
    """總入口：解讀一句話 → 依動作呼叫對應功能 → 回傳給使用者的文字"""
    text = (text or "").strip()
    student = (student or "").strip()
    if not text:
        return "請輸入您的需求，例如：我要訂兩杯拿鐵。"

    action, item, qty, raw = interpret(text)

    if action == "統計":                      # 統計不需要姓名
        return get_sales()
    if action == "未知" or item == "無":
        return unknown_reply()
    if action == "查詢":                      # 查詢不需要姓名
        return query_stock(item)

    # 訂購與取消需要姓名
    if not student:
        return "⚠️ 訂購或取消需要輸入姓名。"
    if action == "訂購":
        return order_drink(student, item, qty)
    if action == "取消":
        return cancel_order(student, item)
    return unknown_reply()


print("✅ 功能函式已定義")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# 流程測試：每一步用 handle_request() 跑一次完整流程，
# 並檢查系統回應有沒有包含「預期關鍵字」。每次都先重建資料庫，確保結果可重現。
init_db()

# 格式：(姓名, 使用者說的話, 預期回應中要出現的關鍵字)
FLOW_TESTS = [
    ("小明", "我要兩杯拿鐵",                 "預約成功"),       # 1 正常訂購（130 元）
    ("小明", "我要訂一杯美式咖啡",           "預約成功"),       # 2 小明累計 3 杯
    ("小明", "我要訂一杯抹茶歐蕾",           "每人上限"),       # 3 超過每人 3 杯，被拒
    ("小美", "給我十杯美式咖啡",             "數量需介於"),     # 4 數量超過 3 杯，由程式拒絕
    ("小美", "我要兩杯拿鐵",                 "預約成功"),       # 5 拿鐵被訂完（4 杯）
    ("小華", "我要兩杯拿鐵",                 "庫存不足"),       # 6 拿鐵剩 0 杯，被拒
    ("小明", "幫我取消拿鐵",                 "取消成功"),       # 7 取消並釋出 2 杯
    ("小華", "幫我取消美式咖啡",             "查無訂單"),       # 8 沒訂過，查不到訂單
    ("",     "抹茶歐蕾還剩幾杯？",           "剩餘 3 杯"),      # 9 查詢（不需要姓名）
    ("",     "今天各品項賣得怎樣",           "今日銷售統計"),   # 10 統計（JOIN + GROUP BY，不需要姓名）
]

flow_passed = 0
for i, (student, text, keyword) in enumerate(FLOW_TESTS, start=1):
    reply = handle_request(text, student)
    ok = keyword in reply
    flow_passed += ok
    print(f"【第 {i} 步】{'✅' if ok else '❌'} 姓名「{student or '（空白）'}」：{text}")
    print(f"  系統回應：{reply}\n")

print(f"🎯 流程測試：通過 {flow_passed} / {len(FLOW_TESTS)} 步\n")
show_tables()
````
